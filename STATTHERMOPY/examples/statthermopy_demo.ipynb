# StatThermoPy — Statistical Thermodynamics Demo

Every property below is derived **only** from the molecular partition function
$Q = Q_t Q_r Q_v Q_e$ — no empirical correlations (NASA/JANAF/Shomate/CoolProp).

```python
import numpy as np
import matplotlib.pyplot as plt
from statthermopy import State, Thermodynamics, IdealGasMixture, get, list_molecules
print(list_molecules())
```

## 1. Pure gas: N2 at 298.15 K, 1 atm

```python
n2 = get('N2')
th = Thermodynamics(n2, State(T=298.15, P=101325.0)).compute()
print(f'Cp_m = {th.Cp_m:.4f} J/mol/K  (lit ~29.10)')
print(f'gamma = {th.gamma:.6f}        (lit ~1.40)')
print(f'S_m  = {th.S_m:.4f} J/mol/K  (lit ~191.5)')
print(f'Q   = {th.Qtotal:.4e}')
```

## 2. Per-mode breakdown for H2O (nonlinear)

```python
h2o = get('H2O')
pf = Thermodynamics(h2o, State(T=300.0, P=1e5)).partition
for name, c in pf.contributions(State(T=300.0, P=1e5)).items():
    print(f'{name:13s}  lnQ={c.ln_q:+.4f}  U={c.U_m:10.3f}  S={c.S_m:8.3f}  Cv={c.Cv_m:7.3f}')
```

## 3. Cp(T) for several gases

```python
T = np.linspace(300, 2000, 100)
fig, ax = plt.subplots(figsize=(8,5))
for name in ['N2', 'O2', 'CO2', 'H2O', 'CH4']:
    mol = get(name)
    Cp = [Thermodynamics(mol, State(T=t, P=1e5)).compute().Cp_m for t in T]
    ax.plot(T, Cp, label=name)
ax.set_xlabel('T (K)'); ax.set_ylabel('Cp_m (J/mol/K)')
ax.set_title('Cp vs T (statistical mechanics)'); ax.legend(); ax.grid(alpha=0.3)
```

## 4. Ideal-gas mixture: CH4 + CO2 (50/50 mole) at 800 K, 5 bar

```python
mix = IdealGasMixture.from_names({'CH4': 0.5, 'CO2': 0.5})
r = mix.compute(State(T=800.0, P=5e5))
print(f'M_avg = {r.M_avg*1e3:.4f} g/mol')
print(f'Cp_m = {r.Cp_m:.4f}  Cv_m = {r.Cv_m:.4f}  gamma = {r.gamma:.6f}')
print(f'S_m  = {r.S_m:.4f} J/mol/K  (includes mixing entropy)')
```
